```python
import pandas as pd

# Read the TSV file with specific columns
columns_to_read = ['rowid', 'Scan', 'Annotation','Score','ProtsAll', 'AnnotationOther', 'ChargeOther','ScoreOther','ProtsAllOther',
                   'MinNTermAdd', 'minNTermSubtract', 'MinCTermAdd', 'minCTermSubtract']
mismatch_df = pd.read_csv('PRISMAL_REPROCESS_PA_UniproKB_Normal123_mismatched.tsv', sep='\t', usecols=columns_to_read)
matched_df = pd.read_csv('PRISMAL_REPROCESS_PA_UniproKB_Normal123_matched.tsv', sep='\t', usecols=columns_to_read)

# Display basic information about the dataset
print(f"Dataset shape: {mismatch_df.shape}")
print(f"Columns: {list(mismatch_df.columns)}")
print("\nFirst few rows:")
mismatch_df
```

```python
# Concatenate matched and mismatch dataframes (add source column to keep track)
combined_df = pd.concat(
    [matched_df.assign(source='matched'), mismatch_df.assign(source='mismatch')],
    ignore_index=True
)

print(f"Combined shape: {combined_df.shape}")
combined_df.head()
```

```python
peptide_set = set()
for annotation, annotation_other in combined_df[['Annotation', 'AnnotationOther']].itertuples(index=False):
    if isinstance(annotation, str) and annotation:
        peptide_set.add(annotation)
    if isinstance(annotation_other, str) and annotation_other:
        peptide_set.add(annotation_other)

len(peptide_set)
```

```python
annotation_counts = combined_df['Annotation'].value_counts()
annotation_other_counts = combined_df['AnnotationOther'].value_counts()

peptide_counts_df = (
    pd.DataFrame({'peptide': sorted(peptide_set)})
    .assign(
        identified_DB=lambda df: df['peptide'].map(annotation_counts).fillna(0).astype(int),
        identified_PA=lambda df: df['peptide'].map(annotation_other_counts).fillna(0).astype(int),
    )
)
peptide_counts_df.to_csv('peptide_counts.tsv', sep='\t', index=False)
```
