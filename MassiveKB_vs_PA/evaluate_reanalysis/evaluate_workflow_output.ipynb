```python
import pandas as pd
import glob

# Read all files matching the pattern "PRISMAL.DB*view-main.tsv"
file_pattern = "PRISMAL.DB*view-main.tsv"
# file_pattern = "USI*.tsv"
files = glob.glob(file_pattern)

# Read and combine all matching files
dataframes = []
for file in files:
    df = pd.read_csv(file, sep='\t')
    dataframes.append(df)

# Combine all dataframes
if dataframes:
    combined_df = pd.concat(dataframes, ignore_index=True)
    print(f"Read {len(files)} files with {len(combined_df)} total rows")
else:
    print("No files found matching the pattern")
```

```python
# Find the maximum score for each scan number
scan_max_score = {} #in format {scan_number: score}
# Get the index of the row with maximum score for each scan
max_score_indices = combined_df.groupby('Scan')['Score'].idxmax()

# Filter to keep only rows where the score equals the maximum score for that scan
filtered_df = combined_df.loc[max_score_indices].fillna('N/A')
# filtered_df['SpectrumFile'] = 'xkan/July_mgfs/Normal_1_2_3.mgf'
if 'SpectrumFileOriginal' in filtered_df.columns:
    filtered_df = filtered_df.drop('SpectrumFileOriginal', axis=1)
    
print(f"Original rows: {len(combined_df)}, Filtered rows: {len(filtered_df)}")
```

```python
filtered_df.to_csv('filtered_combined_psms.tsv', sep='\t', index=False)
# filtered_df.to_csv('precursor_filtered_psms.tsv', sep='\t', index=False)
print(f"Saved {len(filtered_df)} rows to filtered_combined_psms.tsv")
```
