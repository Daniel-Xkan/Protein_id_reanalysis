```python
import csv

usi_dict = {}
with open('all_usi.tsv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter='\t')
    for row in reader:
        usi = row['USI']
        important = '1' if row.get('important') == '1' else '0'
        usi_dict[usi] = important
```

```python
# To view the first few items in the dictionary, use list() and slicing
list(usi_dict.items())[:5]
```

read in from all the usis_xxxx.txt (does not have keyword important in file name), out put the filtered usis_xxxx_important.txt whose values of important ==1 in the previous step

```python
import glob

for file in glob.glob('usis_*.txt'):
    if 'important' in file:
        continue

    filtered_rows = []
    with open(file, 'r', encoding='utf-8') as fin:
        header = fin.readline()
        filtered_rows.append(header)
        for line in fin:
            usi = line.strip().split('\t')[0]
            # Debug print
            print(f"{usi}: {usi_dict.get(usi)}")
            if usi_dict.get(usi) == '1':
                filtered_rows.append(line)

    out_file = file.replace('.txt', '_important.txt')
    with open(out_file, 'w', encoding='utf-8') as fout:
        fout.writelines(filtered_rows)
```
