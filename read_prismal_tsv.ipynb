# Reading PRISMAL-REPROCESS_PA_UniproKB_Normal123.tsv

This notebook demonstrates how to read and analyze the PRISMAL-REPROCESS TSV file.

```python
import pandas as pd
import os
```

## Load the TSV File

```python
# File path
file_path = 'PRISMAL-REPROCESS_PA_UniproKB_Normal123.tsv'

# Read the TSV file
df = pd.read_csv(file_path, sep='\t', low_memory=False)

print(f"Successfully loaded {len(df)} rows and {len(df.columns)} columns")
```

## Examine the Data

```python
# Display basic information
print("Shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
# Display first few rows
df.head()
```

## Basic Statistics

```python
# Statistical summary of numeric columns
df.describe()
```

## Additional Analysis Examples

```python
# Example: Filter high confidence results (if Confidence_Score column exists)
if 'Confidence_Score' in df.columns:
    high_confidence = df[df['Confidence_Score'] > 0.9]
    print(f"High confidence entries (>0.9): {len(high_confidence)}")
    print(high_confidence)
```

```python
# Example: Check for unique proteins (if Protein_ID column exists)
if 'Protein_ID' in df.columns:
    unique_proteins = df['Protein_ID'].nunique()
    print(f"Number of unique proteins: {unique_proteins}")
```
