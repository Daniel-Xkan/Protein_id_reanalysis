```python
# Read the file and extract all modifications from USI strings
with open('../new_usi/all_usi.txt', 'r') as f:
    lines = f.readlines()

# Set to store unique modifications
mods = set()

# Process each line to find modifications in []
for line in lines:
    line = line.strip()
    if '[' in line and ']' in line:
        # Find all modifications in square brackets
        start = 0
        while True:
            start_idx = line.find('[', start)
            if start_idx == -1:
                break
            end_idx = line.find(']', start_idx)
            if end_idx == -1:
                break
            mod = line[start_idx:end_idx+1]
            mods.add(mod)
            start = end_idx + 1

# Write summary to file
with open('modifications_summary.txt', 'w') as f:
    f.write(f"Total unique modifications found: {len(mods)}\n\n")
    f.write("All modifications:\n")
    for mod in sorted(mods):
        f.write(f"{mod}\n")

print(f"Found {len(mods)} unique modifications")
print(f"Summary written to modifications_summary.txt")
```
